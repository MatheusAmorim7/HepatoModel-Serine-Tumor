# Hepatic serine supply and tumour demand
Run the cells in order. The repository contains the original metabolic model code and the inputs required to reproduce poster Figures 2 and 3. These are separate well-mixed tests, not a spatial tumour simulation or a calibrated growth model.
Outputs are saved in a new `results_colab_...` folder within the cloned repository. Colab storage is temporary; download results before ending the session.

```python
from pathlib import Path
from datetime import datetime
import subprocess
import sys

ROOT = Path('/content/HepatoModel-Serine-Tumor')
if not ROOT.exists():
    subprocess.run(['git', 'clone', 'https://github.com/MatheusAmorim7/HepatoModel-Serine-Tumor.git', str(ROOT)], check=True)
subprocess.run([sys.executable, '-m', 'pip', 'install', '-r', str(ROOT / 'requirements.txt')], check=True)
```

```python
OUT = ROOT / ('results_colab_' + datetime.now().strftime('%Y%m%d_%H%M%S'))
subprocess.run([sys.executable, str(ROOT / 'reproduce.py'), '--out-dir', str(OUT)], cwd=ROOT, check=True)
print('Results:', OUT)
```

```python
from IPython.display import Image, display
display(Image(filename=str(OUT / 'figure2_demand.png')))
display(Image(filename=str(OUT / 'figure3_supply_HPRT.png')))
print((OUT / 'checks.json').read_text())
```
